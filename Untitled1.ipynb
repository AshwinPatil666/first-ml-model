```python
import pandas as pd
import numpy as np

# Random seed fix karne se data repeatable hoga
np.random.seed(42)

# 100 students ke liye data generate
data = {
    "cgpa": np.round(np.random.uniform(5.0, 10.0, 100), 2),   # CGPA between 5.0 - 10.0
    "iq": np.random.randint(80, 150, 100),                    # IQ between 80 - 150
    "placement": np.random.choice(["1", "0"], 100)         # Placement status Yes/No
}

# DataFrame banaye
df = pd.DataFrame(data)
```
